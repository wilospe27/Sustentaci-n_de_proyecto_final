```python
#Paso 1.1: Elijo las columnas de interés agendamiento.xlsx

columns = [
    'Id_med_paciente', 'sexo', 'nivel_usr', 'tipo_usuario',
    'edad', 'nombre_cnt', 'turno', 'fecha_cita', 'fecha_asig',
    'control', 'especialidad_cita'
]

agendamiento = agendamiento.loc[:, columns]
```

```python
agendamiento
```

```python
#Paso 2: Cambio los tipo de datos de las columnas
#Id_med_paciente: Str // sexo: Str // nivel_usr: int // tipo_usuario: str // Edad: Int //
#nombre_cnt: Str // Turno: Hora (llevar a rango hora 7, 8, 9) // Fecha cita y fecha_asig : Datatime
# Control: X=1, Missing Value:0 //especialidad_cita: str.

# Paso 2: Cambio los tipo de datos de las columnas
# Id_med_paciente: Str // sexo: Str // nivel_usr: int // tipo_usuario: str // Edad: Int //
# nombre_cnt: Str // Turno: Hora (llevar a rango hora 7, 8, 9) // Fecha cita y fecha_asig : Datatime
# Control: X=1, Missing Value:0 // especialidad_cita: str.

agendamiento["Id_med_paciente"] = agendamiento["Id_med_paciente"].astype(str)
agendamiento["sexo"] = agendamiento["sexo"].astype(str)
agendamiento["nivel_usr"] = agendamiento["nivel_usr"].astype("Int64")
agendamiento["tipo_usuario"] = agendamiento["tipo_usuario"].astype(str)
agendamiento["edad"] = agendamiento["edad"].astype("Int64")
agendamiento["nombre_cnt"] = agendamiento["nombre_cnt"].astype(str)
agendamiento["fecha_cita"] = pd.to_datetime(agendamiento["fecha_cita"], errors="coerce")
agendamiento["fecha_asig"] = pd.to_datetime(agendamiento["fecha_asig"], errors="coerce")

# Turno: convertir texto de hora a un valor numérico por franja horaria
agendamiento["turno"] = pd.to_datetime(agendamiento["turno"], format="%H:%M", errors="coerce")
minutos = agendamiento["turno"].dt.hour * 60 + agendamiento["turno"].dt.minute
agendamiento['turno_sin'] = np.sin(2 * np.pi * minutos / 1440)
agendamiento['turno_cos'] = np.cos(2 * np.pi * minutos / 1440)

# Control: X = 1, faltante = 0
agendamiento["control"] = agendamiento["control"].map({"X": 1}).fillna(0).astype(int)

# Especialidad: texto
agendamiento["especialidad_cita"] = agendamiento["especialidad_cita"].astype(str)
```

```python
#Construyo una columna de lead time, en donde identifico el tiempo transcurrido entre fecha_asign hasta fecha_cita
agendamiento["lead_time"] = (
    agendamiento["fecha_cita"] - agendamiento["fecha_asig"]
).dt.days
```

```python
#Llenamos Missing value en programa con Programa de Hipertension, valor que ya se encuentra en la columna# Llenamos los valores faltantes de Programa
rcv["Programa"] = rcv["Programa"].fillna("Programa de Hipertension")

# Columna diabetes cambios valores x=1 y Missing Value=0
# Diabetes: X = 1, valor faltante = 0
rcv["Diabetes"] = rcv["Diabetes"].eq("X").astype(int)

display(rcv)
```

```python
#quiero ver las categorías de la columna diabetes por cantidad y por porcentaje 
diabetes_counts = rcv["Diabetes"].value_counts().sort_index()
diabetes_pct = rcv["Diabetes"].value_counts(normalize=True).sort_index() * 100

pd.DataFrame({
    "cantidad": diabetes_counts,
    "porcentaje": diabetes_pct
})
```
